```python
from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    pipeline,
)
```

```python
model_name = "Qwen/Qwen2.5-1.5B-Instruct"
```

```python
tokenizer = AutoTokenizer.from_pretrained(model_name)
```

```python
model = AutoModelForCausalLM.from_pretrained(
    model_name,
    device_map="auto"
)
```

```python
pipe = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer
)
```

```python
prompt = "Patient has fever for three days. What could be the possible causes?"

response = pipe(
    prompt,
    max_new_tokens=150,
)

print(response[0]["generated_text"])
```

```python
print("AI Healthcare Assistant")
print("Type 'exit' to quit.\n")

while True:
    user_input = input("You: ")

    if user_input.lower() == "exit":
        print("Assistant: Goodbye! Stay healthy.")
        break

    response = pipe(
        user_input,
        max_new_tokens=150,
        temperature=0.7,
        do_sample=True
    )

    print("\nAssistant:")
    print(response[0]["generated_text"])
    print("-" * 60)
```
