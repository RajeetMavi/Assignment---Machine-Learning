# Strengths and Weakneses

## Meta Llama 3.1 8B Instruct

### Strengths

- Excellent instruction following
- Strong reasoning ability
- Produces detailed and coherent responses
- Performs well on healthcare, coding, and analytical tasks

### Weaknesses

- Requires more computational resources
- Larger download size
- Slower than smaller models


## Qwen2.5-1.5B-Instruct

### Strengths

- Lightweight and efficient
- Faster inference
- Runs comfortably on free Google Colab GPUs
- Good instruction following despite its smaller size
- Strong multilingual capabilities

### Weaknesses

- Less accurate than larger models on complex reasoning
- May generate shorter or less detailed answers
- Performance decreases on tasks requiring extensive contextual understanding

```python
from huggingface_hub import login

login()
```

```python
from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    pipeline,
)
```

```python
tokenizer_Qwen = AutoTokenizer.from_pretrained("Qwen/Qwen2.5-1.5B-Instruct")
```

```python
tokenizer_llama = AutoTokenizer.from_pretrained("meta-llama/Llama-3.1-8B")
```

```python
model_1 = AutoModelForCausalLM.from_pretrained(
    "Qwen/Qwen2.5-1.5B-Instruct",
    device_map="auto"
)
```

```python
model_2 = AutoModelForCausalLM.from_pretrained(
    "meta-llama/Llama-3.1-8B",
    device_map="auto"
)
```

```python
llama_pipe = pipeline(
    "text-generation",
    model=model_2,
    tokenizer=tokenizer_llama
)
```

```python
Qwen_pipe = pipeline(
    "text-generation",
    model=model_1,
    tokenizer=tokenizer_Qwen
)
```

```python
question = "What is diabetes?"
```

```python
llama_response = llama_pipe(
    question,
    max_new_tokens=150,
)

print("Lamma Response: ")
print(llama_response[0]["generated_text"])
```

```python
Qwen_response = Qwen_pipe(
    question,
    max_new_tokens=150,
)

print("Qwen Response: ")
print(Qwen_response[0]["generated_text"])
```
