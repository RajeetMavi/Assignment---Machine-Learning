# Task 7 — Prompt Optimization

The AI currently produces the following response:

Maybe the patient has malaria.
Or maybe dengue.
Or maybe COVID.
I am not sure.

You need to answer three questions:

Why is the response poor?
Which Prompt Engineering techniques would improve it?
Rewrite the prompt to reduce hallucinations.

# Task 7 - Solution

## Why is the Response Poor?

The response has several issues:

- Hallucination: The AI may be hallucinating and hence it suggests diseases (malaria, dengue, COVID) without sufficient evidence.
- Bad Prompting: May be the user had provide a one line prompt, not stating clearly what the user wants.
- No structured output: The answer is disorganized and difficult to interpret.
- No recommendation: The model does not advise the user on appropriate next steps, such as seeking medical evaluation.
- High uncertainty: The phrase "I am not sure" is highly uncertain and does not provide any useful guidance or clarfying that more info is needed.

## Which Prompt Engineering techniques would improve it?

To improve it, I would suggest to use hybrid prompting technique, using role prompting, context and CoT to give a proper structured answer.

## Rewriting  the prompt to reduce hallucinations.

Role:
You are an AI healthcare assistant.

Patient Information:
- Fever for 3 days
- Cough
